```python
import cv2
import mediapipe as mp
import matplotlib.pyplot as plt

# Load video
video_path = '/drive/MyDrive/cricket_clip.mp4'
cap = cv2.VideoCapture(video_path)

# MediaPipe pose model
mp_pose = mp.solutions.pose
pose = mp_pose.Pose()
mp_drawing = mp.solutions.drawing_utils

frame_count = 0

while cap.isOpened() and frame_count < 10:  # Just first 10 frames for demo
    success, frame = cap.read()
    if not success:
        break

    image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = pose.process(image_rgb)

    if results.pose_landmarks:
        mp_drawing.draw_landmarks(frame, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)

    # Convert BGR to RGB for matplotlib
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Show using matplotlib
    plt.figure(figsize=(10, 6))
    plt.imshow(frame_rgb)
    plt.axis('off')
    plt.title(f"Frame {frame_count}")
    plt.show()

    frame_count += 1

cap.release()
```

```python
import cv2
import mediapipe as mp

# Initialize mediapipe pose
mp_pose = mp.solutions.pose
pose = mp_pose.Pose()
mp_drawing = mp.solutions.drawing_utils

# Load video (replace with your file)
cap = cv2.VideoCapture(video_path)

# Define landmark names
landmark_names = mp_pose.PoseLandmark

frame_count = 0

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    frame_count += 1

    # Convert to RGB
    image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = pose.process(image)

    # Convert back to BGR for OpenCV
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

    if results.pose_landmarks:
        # Draw the skeleton
        mp_drawing.draw_landmarks(
            image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)

        # Draw text labels
        for idx, landmark in enumerate(results.pose_landmarks.landmark):
            h, w, _ = image.shape
            cx, cy = int(landmark.x * w), int(landmark.y * h)
            landmark_name = landmark_names(idx).name
            cv2.putText(image, f'{idx}:{landmark_name}', (cx, cy),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0, 255, 255), 1)

    # Save a few sample frames
    if frame_count <= 5:
        cv2.imwrite(f"pose_frame_{frame_count}.png", image)

cap.release()
```
